```python
import pandas as pd
```

```python
df = pd.read_csv("Churn_Modelling.csv")
```

```python
print(df.shape)
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
df = df.drop(columns=["RowNumber", "CustomerId", "Surname"], errors="ignore")
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df["Status"] = df["Exited"].map({1: "Churned", 0: "Retained"})
df["ActiveStatus"] = df["IsActiveMember"].map({1: "Active", 0: "Inactive"})
df["HasCard"] = df["HasCrCard"].map({1: "Yes", 0: "No"})
```

```python
df["AgeGroup"] = pd.cut(df["Age"],
    bins=[0, 30, 40, 50, 60, 120],
    labels=["Under 30", "30-40", "40-50", "50-60", "60+"],
    right=False)
```

```python
print(df["Status"].value_counts())
```

```python
print(df.groupby("Geography")["Exited"].mean())
```

```python
df.to_csv("churn_cleaned.csv", index=False)
```
